```python
# ==============================================================
# 🧑‍🔬 STEP 2B: LOAD FACE + AGE/GENDER MODELS
# ==============================================================
print('=' * 70)
print('🧑‍🔬 LOADING FACE + AGE/GENDER MODELS')
print('=' * 70)

import os
import urllib.request
import cv2
import numpy as np

model_dir = '/content/models'
os.makedirs(model_dir, exist_ok=True)

def download_if_missing(url, path):
    if not os.path.exists(path):
        print(f'Downloading {os.path.basename(path)}...')
        urllib.request.urlretrieve(url, path)

# Face detector (OpenCV DNN)
face_proto = os.path.join(model_dir, 'deploy.prototxt')
face_model = os.path.join(model_dir, 'res10_300x300_ssd_iter_140000_fp16.caffemodel')
download_if_missing('https://raw.githubusercontent.com/opencv/opencv/master/samples/dnn/face_detector/deploy.prototxt', face_proto)
download_if_missing('https://raw.githubusercontent.com/opencv/opencv_3rdparty/master/dnn_samples_face_detector_20170830/res10_300x300_ssd_iter_140000_fp16.caffemodel', face_model)

# Age/Gender models (Caffe)
age_proto = os.path.join(model_dir, 'age_deploy.prototxt')
age_model = os.path.join(model_dir, 'age_net.caffemodel')
gender_proto = os.path.join(model_dir, 'gender_deploy.prototxt')
gender_model = os.path.join(model_dir, 'gender_net.caffemodel')
download_if_missing('https://raw.githubusercontent.com/spmallick/learnopencv/master/AgeGender/age_deploy.prototxt', age_proto)
download_if_missing('https://raw.githubusercontent.com/spmallick/learnopencv/master/AgeGender/age_net.caffemodel', age_model)
download_if_missing('https://raw.githubusercontent.com/spmallick/learnopencv/master/AgeGender/gender_deploy.prototxt', gender_proto)
download_if_missing('https://raw.githubusercontent.com/spmallick/learnopencv/master/AgeGender/gender_net.caffemodel', gender_model)

face_net = cv2.dnn.readNetFromCaffe(face_proto, face_model)
age_net = cv2.dnn.readNetFromCaffe(age_proto, age_model)
gender_net = cv2.dnn.readNetFromCaffe(gender_proto, gender_model)

AGE_LIST = ['(0-2)','(4-6)','(8-12)','(15-20)','(25-32)','(38-43)','(48-53)','(60-100)']
GENDER_LIST = ['Male','Female']
MODEL_MEAN_VALUES = (78.4263377603, 87.7689143744, 114.895847746)

def detect_faces(frame, conf_threshold=0.7):
    h, w = frame.shape[:2]
    blob = cv2.dnn.blobFromImage(frame, 1.0, (300, 300), (104.0, 177.0, 123.0))
    face_net.setInput(blob)
    detections = face_net.forward()
    boxes = []
    for i in range(detections.shape[2]):
        confidence = detections[0, 0, i, 2]
        if confidence > conf_threshold:
            box = detections[0, 0, i, 3:7] * np.array([w, h, w, h])
            x1, y1, x2, y2 = box.astype(int)
            x1, y1 = max(0, x1), max(0, y1)
            x2, y2 = min(w-1, x2), min(h-1, y2)
            if x2 > x1 and y2 > y1:
                boxes.append((x1, y1, x2, y2))
    return boxes

def estimate_age_gender(face_img):
    blob = cv2.dnn.blobFromImage(face_img, 1.0, (227, 227), MODEL_MEAN_VALUES, swapRB=False)
    gender_net.setInput(blob)
    gender_preds = gender_net.forward()
    gender = GENDER_LIST[gender_preds[0].argmax()]

    age_net.setInput(blob)
    age_preds = age_net.forward()
    age = AGE_LIST[age_preds[0].argmax()]
    return age, gender

print('✅ Face + age/gender models ready!')
```

```python
# ==============================================================
# 🎬 STEP 5: PROCESS VIDEO - DETECT, TRACK & TIMESTAMP
# ==============================================================
print("=" * 70)
print("🚀 PROCESSING VIDEO")
print("=" * 70)

import cv2
import time
import pandas as pd
import numpy as np
from datetime import datetime

# Open video
cap = cv2.VideoCapture(video_path)

# Get video properties
width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS)) or 25
total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

print(f"Video Information:")
print(f"   - Resolution: {width}x{height}")
print(f"   - FPS: {fps}")
print(f"   - Total Frames: {total_frames}")
print(f"   - Duration: {total_frames/fps:.1f} seconds\n")

# Setup output video
output_path = "/content/output_tracked.mp4"
fourcc = cv2.VideoWriter_fourcc(*'mp4v')
out = cv2.VideoWriter(output_path, fourcc, fps, (width, height))

# Data storage for tracking
track_data = {}  # Store: {track_id: {"start_frame": frame, "end_frame": frame, "age": str, "gender": str}}
frame_count = 0

# Heatmap accumulator
heatmap = np.zeros((height, width), dtype=np.float32)

# Helper function to format duration
def format_duration(seconds):
    """Convert seconds to readable format: Xh Ym Zs or Ym Zs or Zs"""
    hours = int(seconds // 3600)
    minutes = int((seconds % 3600) // 60)
    secs = int(seconds % 60)

    if hours > 0:
        return f"{hours}h {minutes}m {secs}s"
    elif minutes > 0:
        return f"{minutes}m {secs}s"
    else:
        return f"{secs}s"

# Helper function to convert frame number to video timestamp
def frame_to_timestamp(frame_num, fps):
    """Convert frame number to actual video time in seconds"""
    return frame_num / fps

print("Starting video processing...")
print("This may take a few minutes depending on video length.\n")

processing_start = time.time()

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    frame_count += 1

    # Show progress every 30 frames
    if frame_count % 30 == 0:
        progress = (frame_count / total_frames) * 100
        elapsed = time.time() - processing_start
        fps_processing = frame_count / elapsed if elapsed > 0 else 0
        eta = (total_frames - frame_count) / fps_processing if fps_processing > 0 else 0

        print(f"Progress: {progress:.1f}% | Frame {frame_count}/{total_frames} | "
              f"Speed: {fps_processing:.1f} fps | ETA: {format_duration(eta)}")

    # ============================================
    # DETECTION: Detect people in current frame
    # ============================================
    results = model.predict(
        frame,
        conf=0.45,       # Increased confidence threshold for better accuracy
        classes=[0],     # Class 0 = person in COCO dataset
        verbose=False,
        iou=0.5         # IoU threshold for NMS
    )

    # Prepare detections for tracker
    detections = []
    if len(results[0].boxes) > 0:
        for box, conf in zip(results[0].boxes.xyxy, results[0].boxes.conf):
            x1, y1, x2, y2 = box.cpu().numpy()
            confidence = float(conf)

            # Ensure bounding box is valid and within frame
            x1, y1 = max(0, int(x1)), max(0, int(y1))
            x2, y2 = min(width, int(x2)), min(height, int(y2))

            # Only add detection if box is valid
            if x2 > x1 and y2 > y1:
                # Format: ([x1, y1, width, height], confidence, class_name)
                detections.append(([x1, y1, x2-x1, y2-y1], confidence, "person"))

    # ============================================
    # TRACKING: Update tracks with new detections
    # ============================================
    tracks = tracker.update_tracks(detections, frame=frame)

    # ============================================
    # VISUALIZATION: Draw bounding boxes & timestamps
    # ============================================
    for track in tracks:
        # Only show confirmed tracks (not tentative)
        if not track.is_confirmed():
            continue

        track_id = track.track_id
        ltrb = track.to_ltrb()  # Get bounding box (left, top, right, bottom)
        x1, y1, x2, y2 = map(int, ltrb)

        # Ensure bounding box stays within frame boundaries
        x1, y1 = max(0, x1), max(0, y1)
        x2, y2 = min(width, x2), min(height, y2)

        # ============================================
        # TIME TRACKING: Record entry and update exit time (FIXED)
        # ============================================
# ============================================
# AGE/GENDER ESTIMATION (every ~30 frames per track)
# ============================================
        if (frame_count - track_data[track_id]["last_age_gender_frame"]) >= 30:
            # Crop person ROI and detect face inside it
            person_roi = frame[y1:y2, x1:x2]
            if person_roi.size > 0:
                faces = detect_faces(person_roi, conf_threshold=0.7)
                if faces:
                    # pick largest face
                    fx1, fy1, fx2, fy2 = max(faces, key=lambda b: (b[2]-b[0])*(b[3]-b[1]))
                    face_img = person_roi[fy1:fy2, fx1:fx2]
                    if face_img.size > 0:
                        age, gender = estimate_age_gender(face_img)
                        track_data[track_id]["age"] = age
                        track_data[track_id]["gender"] = gender
                        track_data[track_id]["last_age_gender_frame"] = frame_count

# ============================================
# HEATMAP: Accumulate center points
# ============================================
        cx = int((x1 + x2) / 2)
        cy = int((y1 + y2) / 2)
        if 0 <= cx < width and 0 <= cy < height:
            heatmap[cy, cx] += 1

        if track_id not in track_data:
            # First time seeing this person - record the frame number
            track_data[track_id] = {
                "start_frame": frame_count,
                "end_frame": frame_count,
                "age": None,
                "gender": None,
                "last_age_gender_frame": -999
            }
        else:
            # Update last seen frame
            track_data[track_id]["end_frame"] = frame_count

        # Calculate duration based on ACTUAL VIDEO TIME (not processing time)
        start_time = frame_to_timestamp(track_data[track_id]["start_frame"], fps)
        end_time = frame_to_timestamp(track_data[track_id]["end_frame"], fps)
        duration = end_time - start_time
        time_label = format_duration(duration)

        # ============================================
        # DRAW ON FRAME
        # ============================================

        # Choose color (green for customers, blue for workers - can be customized later)
        color = (0, 255, 0)  # Green in BGR

        # Draw bounding box with thicker line for better visibility
        cv2.rectangle(frame, (x1, y1), (x2, y2), color, 3)

        # Prepare label text
        age_text = track_data[track_id].get("age")
        gender_text = track_data[track_id].get("gender")
        if age_text and gender_text:
            label = f"ID {track_id} | {time_label} | {gender_text}, {age_text}"
        else:
            label = f"ID {track_id} | {time_label}"

        # Calculate label background size
        label_size, baseline = cv2.getTextSize(
            label,
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            2
        )

        # Ensure label stays within frame
        label_y = max(label_size[1] + 10, y1)

        # Draw label background (filled rectangle)
        cv2.rectangle(
            frame,
            (x1, label_y - label_size[1] - 10),
            (x1 + label_size[0] + 10, label_y),
            color,
            -1  # Filled
        )

        # Draw label text
        cv2.putText(
            frame,
            label,
            (x1 + 5, label_y - 5),
            cv2.FONT_HERSHEY_SIMPLEX,
            0.7,
            (255, 255, 255),  # White text
            2
        )

    # Write frame to output video
    out.write(frame)

# Release resources
cap.release()
out.release()

processing_time = time.time() - processing_start

print("\n" + "=" * 70)
print("✅ VIDEO PROCESSING COMPLETE!")
print("=" * 70)
print(f"⏱️  Processing time: {format_duration(processing_time)}")
print(f"🎬 Total frames processed: {frame_count}")
print(f"👥 Unique people detected: {len(track_data)}")
print(f"📁 Output saved to: {output_path}\n")

# ==============================================================
# 📊 OPTIONAL: Print tracking summary
# ==============================================================
print("\n📊 TRACKING SUMMARY:")
print("=" * 70)
for track_id, data in sorted(track_data.items()):
    start_sec = frame_to_timestamp(data["start_frame"], fps)
    end_sec = frame_to_timestamp(data["end_frame"], fps)
    duration = end_sec - start_sec
    print(f"Person ID {track_id}: {format_duration(duration)} "
          f"(from {format_duration(start_sec)} to {format_duration(end_sec)})")
print("=" * 70)
```

```python
# ==============================================================
# 📊 STEP 6: GENERATE ANALYTICS REPORT
# ==============================================================
print("=" * 70)
print("📊 GENERATING ANALYTICS REPORT")
print("=" * 70)

analytics_data = []

for track_id, data in track_data.items():
    start_frame = data["start_frame"]
    end_frame = data["end_frame"]

    # Calculate duration in seconds
    start_sec = frame_to_timestamp(start_frame, fps)
    end_sec = frame_to_timestamp(end_frame, fps)
    duration_seconds = end_sec - start_sec

    analytics_data.append({
        "Person_ID": track_id,
        "First_Seen_Frame": start_frame,
        "Duration_Seconds": round(duration_seconds, 2),
        "Duration_Formatted": format_duration(duration_seconds),
        "Entry_Time": format_duration(start_sec),
        "Exit_Time": format_duration(end_sec),
        "Gender": track_data[track_id].get("gender"),
        "Age_Group": track_data[track_id].get("age"),
    })

# Sort by duration (longest visits first)
analytics_data.sort(key=lambda x: x["Duration_Seconds"], reverse=True)

# Create DataFrame
df = pd.DataFrame(analytics_data)

# Save to CSV
csv_path = "/content/tracking_analytics.csv"
df.to_csv(csv_path, index=False)

# Display summary statistics
print("\n📈 SUMMARY STATISTICS")
print("=" * 70)
print(f"Total People Detected:     {len(track_data)}")
print(f"Average Visit Duration:    {df['Duration_Seconds'].mean():.1f} seconds ({format_duration(df['Duration_Seconds'].mean())})")
print(f"Longest Visit:             {df['Duration_Seconds'].max():.1f} seconds ({df['Duration_Formatted'].iloc[0]})")
print(f"Shortest Visit:            {df['Duration_Seconds'].min():.1f} seconds ({format_duration(df['Duration_Seconds'].min())})")

# Show top 5 longest visits
print("\n🏆 TOP 5 LONGEST VISITS")
print("=" * 70)
print(df.head(5)[['Person_ID', 'Duration_Formatted', 'Entry_Time', 'Exit_Time']].to_string(index=False))

print("\n✅ Analytics CSV saved to: {}\n".format(csv_path))
```

```python
#==============================================================
# CELL 10: ADVANCED THEFT DETECTION - BEHAVIORAL ANALYSIS
# ==============================================================
print("=" * 70)
print("🔍 PROCESSING VIDEO FOR ADVANCED THEFT DETECTION")
print("=" * 70)

import cv2 # Import OpenCV
import time # Import time for processing metrics
from deep_sort_realtime.deepsort_tracker import DeepSort # Import DeepSort
from ultralytics import YOLO # Import YOLO for model definition
from datetime import datetime # Import datetime
from collections import defaultdict, deque
import math
import numpy as np

# Helper function to format duration
def format_duration(seconds):
    """Convert seconds to readable format: Xh Ym Zs or Ym Zs or Zs"""
    hours = int(seconds // 3600)
    minutes = int((seconds % 3600) // 60)
    secs = int(seconds % 60)

    if hours > 0:
        return f"{hours}h {minutes}m {secs}s"
    elif minutes > 0:
        return f"{minutes}m {secs}s"
    else:
        return f"{secs}s"


# Re-load the YOLO model if it's not defined, or ensure it's accessible
# This assumes yolov8n.pt was downloaded by an earlier step.
if 'model' not in locals():
    print("YOLO model not found, loading now...")
    model = YOLO("yolov8n.pt")
    print("YOLO model loaded successfully for theft detection!")

cap_theft = cv2.VideoCapture(theft_video_path)

width_t = int(cap_theft.get(cv2.CAP_PROP_FRAME_WIDTH))
height_t = int(cap_theft.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps_t = int(cap_theft.get(cv2.CAP_PROP_FPS)) or 25
total_frames_t = int(cap_theft.get(cv2.CAP_PROP_FRAME_COUNT))

print(f"Video Information:")
print(f"   - Resolution: {width_t}x{height_t}")
print(f"   - FPS: {fps_t}")
print(f"   - Total Frames: {total_frames_t}")
print(f"   - Duration: {total_frames_t/fps_t:.1f} seconds\n")

output_theft_path = "/content/output_theft_detection.mp4"
fourcc_t = cv2.VideoWriter_fourcc(*'mp4v')
out_theft = cv2.VideoWriter(output_theft_path, fourcc_t, fps_t, (width_t, height_t))

# Reset tracker for theft detection
tracker_theft = DeepSort(
    max_age=30,
    n_init=3,
    max_iou_distance=0.7,
    nn_budget=100,
    embedder="mobilenet"
)

# Enhanced behavioral tracking data
behavior_data = defaultdict(lambda: {
    'positions': deque(maxlen=fps_t * 10),  # 10 seconds of positions
    'bboxes': deque(maxlen=fps_t * 5),      # 5 seconds of bboxes
    'hand_positions': deque(maxlen=fps_t * 3),  # 3 seconds of hand tracking
    'entry_frame': None,
    'exit_frame': None,
    'risk_score': 0,
    'alerts': [],
    'zone_dwell_times': defaultdict(float),
    'last_zone': None,
    'zone_entry_time': None,
    'direction_changes': 0,
    'last_direction': None,
    'speeds': deque(maxlen=fps_t * 3),
    'suspicious_poses': 0,
    'age': None,
    'gender': None,
    'last_age_gender_frame': -999,

    # NEW: Advanced tracking
    'hand_movements': deque(maxlen=fps_t * 5),
    'concealment_events': [],
    'staff_avoidance_count': 0,
    'backward_looks': 0,
    'near_exit_time': 0,
    'bbox_area_history': deque(maxlen=fps_t * 3),
    'posture_changes': deque(maxlen=fps_t * 5),
    'last_significant_move': None,
    'stationary_periods': [],
    'item_interaction_zones': [],
})

frame_count_t = 0
alert_log = []
high_risk_snapshots = []

# Define zones (customize based on your shop layout)
ZONES = {
    'entrance': {'y_range': (0, height_t * 0.2), 'name': 'ENTRANCE'},
    'checkout': {'y_range': (height_t * 0.8, height_t), 'name': 'CHECKOUT'},
    'exit': {'x_range': (width_t * 0.8, width_t), 'name': 'EXIT'},
    'aisles': {'y_range': (height_t * 0.2, height_t * 0.8), 'name': 'AISLES'},
}

def detect_hand_to_bag_motion(person_data, current_bbox):
    """Detect if person is making hand-to-bag motion"""
    if len(person_data['hand_movements']) < fps_t * 2:  # Need 2 seconds of data
        return False

    x1, y1, x2, y2 = current_bbox
    bbox_width = x2 - x1
    bbox_height = y2 - y1

    # Check for rapid hand movements in lower body area (bag area)
    # This is approximated by checking bbox aspect ratio changes
    recent_bboxes = list(person_data['bboxes'])[-fps_t:]
    if len(recent_bboxes) < fps_t // 2:
        return False

    # Calculate area variance in lower half
    lower_area_variance = 0
    for i in range(len(recent_bboxes) - 1):
        w1, h1 = recent_bboxes[i]
        w2, h2 = recent_bboxes[i + 1]
        area_diff = abs((w1 * h1) - (w2 * h2))
        lower_area_variance += area_diff

    # If significant variance in lower body area, likely hand-to-bag motion
    if lower_area_variance > (bbox_width * bbox_height * 0.15):
        return True

    return False

def detect_concealment(person_data, current_bbox):
    """Detect concealment actions (putting items in pockets/bags)"""
    if len(person_data['bbox_area_history']) < fps_t:
        return False

    x1, y1, x2, y2 = current_bbox
    current_area = (x2 - x1) * (y2 - y1)

    # Track sudden changes in body silhouette
    areas = list(person_data['bbox_area_history'])
    if len(areas) < 10:
        return False

    # Check for bulging or expansion in body area
    recent_avg = np.mean(areas[-fps_t//2:])
    earlier_avg = np.mean(areas[:fps_t//2])

    # If recent area is significantly larger, might be concealing items
    if recent_avg > earlier_avg * 1.15:
        return True

    return False

def detect_staff_avoidance(person_pos, all_persons):
    """Detect if person is avoiding staff/other people"""
    if len(all_persons) < 2:
        return False

    px, py = person_pos

    # Check distance to other people
    for other_id, other_data in all_persons.items():
        if len(other_data['positions']) == 0:
            continue

        ox, oy, _ = other_data['positions'][-1]
        distance = math.sqrt((px - ox)**2 + (py - oy)**2)

        # If person suddenly changes direction when near others
        if distance < THEFT_CONFIG['staff_proximity_threshold']:
            return True

    return False

def detect_nervous_behavior(person_data):
    """Detect nervous behavior from movement patterns"""
    if len(person_data['positions']) < fps_t * 3:
        return False

    positions = list(person_data['positions'])

    # Calculate movement jerkiness
    velocities = []
    for i in range(len(positions) - 1):
        dx = positions[i+1][0] - positions[i][0]
        dy = positions[i+1][1] - positions[i][1]
        velocity = math.sqrt(dx**2 + dy**2)
        velocities.append(velocity)

    if len(velocities) < 5:
        return False

    # High variance in velocity = nervous/jittery movement
    velocity_variance = np.var(velocities)
    mean_velocity = np.mean(velocities)

    if mean_velocity > 0 and velocity_variance / mean_velocity > 0.8:
        return True

    return False

print("Starting advanced theft detection analysis...")
print("Monitoring: Hand movements, concealment, staff avoidance, nervous behavior...\n")

processing_start_t = time.time()

while cap_theft.isOpened():
    ret, frame = cap_theft.read()
    if not ret:
        break

    frame_count_t += 1
    current_frame_time = frame_count_t / fps_t

    if frame_count_t % 30 == 0:
        progress = (frame_count_t / total_frames_t) * 100
        elapsed = time.time() - processing_start_t
        fps_processing = frame_count_t / elapsed if elapsed > 0 else 0
        eta = (total_frames_t - frame_count_t) / fps_processing if fps_processing > 0 else 0

        print(f"Progress: {progress:.1f}% | Frame {frame_count_t}/{total_frames_t} | "
              f"Speed: {fps_processing:.1f} fps | ETA: {format_duration(eta)}")

    # ============================================
    # DETECTION
    # ============================================
    results = model.predict(
        frame,
        conf=0.45,
        classes=[0],
        verbose=False,
        iou=0.5
    )

    detections = []
    if len(results[0].boxes) > 0:
        for box, conf in zip(results[0].boxes.xyxy, results[0].boxes.conf):
            x1, y1, x2, y2 = box.cpu().numpy()
            confidence = float(conf)

            x1, y1 = max(0, int(x1)), max(0, int(y1))
            x2, y2 = min(width_t, int(x2)), min(height_t, int(y2))

            if x2 > x1 and y2 > y1:
                detections.append(([x1, y1, x2-x1, y2-y1], confidence, "person"))

    # ============================================
    # TRACKING
    # ============================================
    tracks = tracker_theft.update_tracks(detections, frame=frame)

    # Store all current positions for staff avoidance detection
    current_all_positions = {}
    for track in tracks:
        if track.is_confirmed():
            ltrb = track.to_ltrb()
            x1, y1, x2, y2 = map(int, ltrb)
            center_x = (x1 + x2) / 2
            center_y = (y1 + y2) / 2
            current_all_positions[track.track_id] = (center_x, center_y)

    for track in tracks:
        if not track.is_confirmed():
            continue

        track_id = track.track_id
        ltrb = track.to_ltrb()
        x1, y1, x2, y2 = map(int, ltrb)

        x1, y1 = max(0, x1), max(0, y1)
        x2, y2 = min(width_t, x2), min(height_t, y2)

        center_x = (x1 + x2) / 2
        center_y = (y1 + y2) / 2
        bbox_width = x2 - x1
        bbox_height = y2 - y1
        bbox_area = bbox_width * bbox_height

        person_data = behavior_data[track_id]

        # Record entry frame
        if person_data['entry_frame'] is None:
            person_data['entry_frame'] = frame_count_t

        person_data['exit_frame'] = frame_count_t

        # ============================================
        # STORE TRACKING DATA
        # ============================================
        person_data['positions'].append((center_x, center_y, current_frame_time))
        person_data['bboxes'].append((bbox_width, bbox_height))
        person_data['bbox_area_history'].append(bbox_area)

        # Store hand position (approximated as upper body center)
        hand_y = y1 + (bbox_height * 0.3)  # Upper 30% of body
        person_data['hand_movements'].append((center_x, hand_y))

        # Reset risk score for this frame
        current_risk = 0
        frame_alerts = []

        # ============================================
        # EXISTING BEHAVIORAL ANALYSIS
        # ============================================

        # 1. LOITERING DETECTION
        if len(person_data['positions']) >= fps_t * 5:
            positions_array = np.array([(p[0], p[1]) for p in person_data['positions']])
            position_variance = np.var(positions_array, axis=0).sum()

            if position_variance < THEFT_CONFIG['zone_pixels']:
                time_span = person_data['positions'][-1][2] - person_data['positions'][0][2]
                if time_span >= THEFT_CONFIG['loitering_threshold']:
                    current_risk += RISK_WEIGHTS['loitering']
                    frame_alerts.append('LOITERING')

        # 2. ERRATIC MOVEMENT DETECTION
        if len(person_data['positions']) >= 3:
            recent_positions = list(person_data['positions'])[-fps_t:]
            if len(recent_positions) >= 3:
                directions = []
                for i in range(len(recent_positions) - 1):
                    dx = recent_positions[i+1][0] - recent_positions[i][0]
                    dy = recent_positions[i+1][1] - recent_positions[i][1]
                    angle = math.atan2(dy, dx)
                    directions.append(angle)

                if len(directions) >= 2:
                    direction_changes = sum(1 for i in range(len(directions)-1)
                                          if abs(directions[i+1] - directions[i]) > math.pi/4)

                    if direction_changes >= THEFT_CONFIG['direction_changes_threshold']:
                        current_risk += RISK_WEIGHTS['erratic_movement']
                        frame_alerts.append('ERRATIC_MOVEMENT')

        # 3. QUICK EXIT DETECTION
        if person_data['entry_frame'] is not None:
            time_in_shop = (frame_count_t - person_data['entry_frame']) / fps_t
            near_exit = (y2 > height_t * 0.8) or (x1 < width_t * 0.1) or (x2 > width_t * 0.9)

            if near_exit and time_in_shop <= THEFT_CONFIG['quick_exit_threshold']:
                current_risk += RISK_WEIGHTS['quick_exit']
                frame_alerts.append('QUICK_EXIT')

            # NEW: Lingering near exit
            if near_exit and time_in_shop > 3:
                person_data['near_exit_time'] += (1 / fps_t)
                if person_data['near_exit_time'] > 5:
                    current_risk += RISK_WEIGHTS['lingering_near_exit']
                    frame_alerts.append('LINGERING_NEAR_EXIT')

        # 4. LOOKING AROUND (HEAD MOVEMENT)
        if len(person_data['bboxes']) >= fps_t * 2:
            aspect_ratios = [w/h for w, h in person_data['bboxes'] if h > 0]
            if len(aspect_ratios) > 5:
                aspect_variance = np.var(aspect_ratios)
                if aspect_variance >= THEFT_CONFIG['looking_around_threshold']:
                    current_risk += RISK_WEIGHTS['looking_around']
                    frame_alerts.append('LOOKING_AROUND')

        # 5. SUSPICIOUS POSE DETECTION
        if len(person_data['bboxes']) >= fps_t:
            recent_heights = [h for w, h in list(person_data['bboxes'])[-fps_t:]]
            avg_height = np.mean(recent_heights)
            current_height = bbox_height

            if current_height < avg_height * THEFT_CONFIG['crouching_height_ratio']:
                current_risk += RISK_WEIGHTS['suspicious_pose']
                frame_alerts.append('SUSPICIOUS_POSE')
                person_data['suspicious_poses'] += 1

        # ============================================
        # NEW: ADVANCED THEFT DETECTION
        # ============================================

        # 6. HAND-TO-BAG MOTION DETECTION
        if detect_hand_to_bag_motion(person_data, (x1, y1, x2, y2)):
            current_risk += RISK_WEIGHTS['hand_to_bag']
            frame_alerts.append('HAND_TO_BAG')
            person_data['concealment_events'].append({
                'frame': frame_count_t,
                'type': 'hand_to_bag',
                'timestamp': current_frame_time
            })

        # 7. CONCEALMENT DETECTION
        if detect_concealment(person_data, (x1, y1, x2, y2)):
            current_risk += RISK_WEIGHTS['concealment_action']
            frame_alerts.append('CONCEALMENT')
            person_data['concealment_events'].append({
                'frame': frame_count_t,
                'type': 'concealment',
                'timestamp': current_frame_time
            })

        # 8. STAFF AVOIDANCE DETECTION
        if detect_staff_avoidance((center_x, center_y), behavior_data):
            current_risk += RISK_WEIGHTS['avoiding_staff']
            frame_alerts.append('AVOIDING_STAFF')
            person_data['staff_avoidance_count'] += 1

        # 9. NERVOUS BEHAVIOR DETECTION
        if detect_nervous_behavior(person_data):
            current_risk += RISK_WEIGHTS['nervous_behavior']
            frame_alerts.append('NERVOUS_BEHAVIOR')

        # 10. HUNCHED POSTURE (Hiding something under clothing)
        if bbox_width > 0 and bbox_height > 0:
            aspect_ratio = bbox_width / bbox_height
            if aspect_ratio > THEFT_CONFIG['hunched_aspect_ratio']:
                current_risk += RISK_WEIGHTS['item_under_clothing']
                frame_alerts.append('HUNCHED_POSTURE')

        # 11. BACKWARD LOOKS (Checking if being watched)
        if len(person_data['positions']) >= 3:
            recent_pos = list(person_data['positions'])[-3:]
            if len(recent_pos) >= 3:
                # Check if person reversed direction (looking back)
                dx1 = recent_pos[1][0] - recent_pos[0][0]
                dx2 = recent_pos[2][0] - recent_pos[1][0]

                if dx1 * dx2 < 0:  # Direction reversal
                    person_data['backward_looks'] += 1
                    if person_data['backward_looks'] >= 3:
                        current_risk += RISK_WEIGHTS['frequent_looking_back']
                        frame_alerts.append('LOOKING_BACK')

        # ============================================
        # UPDATE RISK SCORE
        # ============================================
        person_data['risk_score'] = max(person_data['risk_score'], current_risk)

        if frame_alerts:
            person_data['alerts'].extend(frame_alerts)

            # Log high-risk events
            if current_risk >= 80:
                alert_entry = {
                    'frame': frame_count_t,
                    'track_id': track_id,
                    'risk_score': current_risk,
                    'alerts': frame_alerts,
                    'timestamp': datetime.now().strftime("%H:%M:%S.%f")[:-3],
                    'concealment_events': len(person_data['concealment_events'])
                }
                alert_log.append(alert_entry)

                # Save snapshot with padding
                snapshot_y1 = max(0, y1-30)
                snapshot_y2 = min(height_t, y2+30)
                snapshot_x1 = max(0, x1-30)
                snapshot_x2 = min(width_t, x2+30)

                snapshot = frame[snapshot_y1:snapshot_y2, snapshot_x1:snapshot_x2].copy()
                high_risk_snapshots.append({
                    'track_id': track_id,
                    'frame': frame_count_t,
                    'snapshot': snapshot,
                    'risk_score': current_risk,
                    'alerts': frame_alerts
                })

        # ============================================
        # ENHANCED VISUALIZATION
        # ============================================
        risk_color = get_risk_color(current_risk)
        risk_label_text = get_risk_label(current_risk)

        # Draw bounding box (thicker for high risk)
        thickness = 4 if current_risk >= 80 else 3 if current_risk >= 60 else 2
        cv2.rectangle(frame, (x1, y1), (x2, y2), risk_color, thickness)

        # Prepare enhanced label
        label_lines = [f"ID{track_id} | Risk:{current_risk}"]
        label_lines.append(risk_label_text)

        # Add top 2 alerts
        if frame_alerts:
            alerts_text = ", ".join(frame_alerts[:2])
            label_lines.append(alerts_text)

        # Add concealment count if any
        if len(person_data['concealment_events']) > 0:
            label_lines.append(f"Concealments: {len(person_data['concealment_events'])}")

        # Calculate label dimensions
        max_label_width = 0
        total_label_height = 10
        line_heights = []

        for line in label_lines:
            (lw, lh), _ = cv2.getTextSize(
                line,
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                2
            )
            max_label_width = max(max_label_width, lw)
            line_heights.append(lh)
            total_label_height += lh + 5

        # Draw label background
        label_y_start = max(y1 - total_label_height - 5, 10)
        cv2.rectangle(
            frame,
            (x1, label_y_start),
            (x1 + max_label_width + 10, label_y_start + total_label_height),
            risk_color,
            -1
        )

        # Draw label text
        current_y = label_y_start + 15
        for line in label_lines:
            cv2.putText(
                frame,
                line,
                (x1 + 5, current_y),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (255, 255, 255),
                2
            )
            current_y += 18

        # Draw movement trail for suspicious individuals
        if current_risk >= 60 and len(person_data['positions']) >= 2:
            points = np.array([(int(p[0]), int(p[1])) for p in person_data['positions']], dtype=np.int32)
            cv2.polylines(frame, [points], False, risk_color, 2)

            # Mark concealment event locations
            for event in person_data['concealment_events']:
                if frame_count_t - event['frame'] < fps_t * 3:  # Show for 3 seconds
                    cv2.circle(frame, (int(center_x), int(center_y)), 15, (0, 0, 255), 3)
                    cv2.putText(frame, "!", (int(center_x)-5, int(center_y)+5),
                              cv2.FONT_HERSHEY_SIMPLEX, 1, (0, 0, 255), 3)

    # Draw frame info
    high_risk_count = sum(1 for t in tracks if t.is_confirmed() and
                         behavior_data[t.track_id]['risk_score'] >= 80)

    info_text = f"Frame: {frame_count_t}/{total_frames_t} | Active: {len(tracks)} | High Risk: {high_risk_count}"
    cv2.rectangle(frame, (5, height_t - 35), (len(info_text) * 12, height_t - 5), (0, 0, 0), -1)
    cv2.putText(frame, info_text, (10, height_t - 15),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)

    out_theft.write(frame)

cap_theft.release()
out_theft.release()

processing_time_t = time.time() - processing_start_t

print("\n" + "=" * 70)
print("✅ ADVANCED THEFT DETECTION COMPLETE!")
print("=" * 70)
print(f"⏱️  Processing time: {format_duration(processing_time_t)}")
print(f"🎬 Total frames analyzed: {frame_count_t}")
print(f"👥 People tracked: {len(behavior_data)}")
print(f"🚨 High-risk alerts: {len(alert_log)}")
print(f"📸 Snapshots saved: {len(high_risk_snapshots)}")
print(f"🎯 Concealment events detected: {sum(len(d['concealment_events']) for d in behavior_data.values())}")
print(f"📁 Output saved to: {output_theft_path}\n")
```

```python
# ==============================================================
# CELL 9: THEFT DETECTION CONFIGURATION
# ==============================================================
print("=" * 70)
print("⚙️  CONFIGURING THEFT DETECTION PARAMETERS")
print("=" * 70)

import numpy as np
from collections import defaultdict, deque
import math

# Theft detection thresholds
THEFT_CONFIG = {
    'loitering_threshold': 15,        # seconds in same area
    'quick_exit_threshold': 5,        # leaves within 5 seconds
    'direction_changes_threshold': 4,  # changes per 10 seconds
    'erratic_speed_variance': 0.5,    # speed variance threshold
    'high_speed_threshold': 2.0,      # m/s near exit
    'looking_around_threshold': 0.3,  # bbox aspect ratio variance
    'zone_pixels': 100,               # pixels to define "same zone"
}

# Risk scoring weights
RISK_WEIGHTS = {
    'loitering': 30,
    'quick_exit': 40,
    'erratic_movement': 20,
    'looking_around': 15,
    'avoiding_checkout': 50,
    'suspicious_pose': 35,
    'high_speed': 25,
}

# Risk level thresholds
RISK_LEVELS = {
    'normal': (0, 30),      # 🟢
    'monitor': (31, 60),    # 🟡
    'suspicious': (61, 80), # 🟠
    'high_risk': (81, 200), # 🔴
}

def get_risk_color(score):
    """Return BGR color based on risk score"""
    if score <= 30:
        return (0, 255, 0)      # Green - Normal
    elif score <= 60:
        return (0, 255, 255)    # Yellow - Monitor
    elif score <= 80:
        return (0, 165, 255)    # Orange - Suspicious
    else:
        return (0, 0, 255)      # Red - High Risk

def get_risk_label(score):
    """Return risk level text"""
    if score <= 30:
        return "🟢 NORMAL"
    elif score <= 60:
        return "🟡 MONITOR"
    elif score <= 80:
        return "🟠 SUSPICIOUS"
    else:
        return "🔴 HIGH RISK"

print("✅ Theft detection parameters configured!")
print(f"   - Loitering threshold: {THEFT_CONFIG['loitering_threshold']}s")
print(f"   - Quick exit threshold: {THEFT_CONFIG['quick_exit_threshold']}s")
print(f"   - Direction change threshold: {THEFT_CONFIG['direction_changes_threshold']}")
print(f"   - Risk levels: Normal(0-30), Monitor(31-60), Suspicious(61-80), High(81+)")
print()
```

```python
# ==============================================================
# CELL 10: THEFT DETECTION - BEHAVIORAL ANALYSIS
# ==============================================================
print("=" * 70)
print("🔍 PROCESSING VIDEO FOR THEFT DETECTION")
print("=" * 70)

cap_theft = cv2.VideoCapture(theft_video_path)

width_t = int(cap_theft.get(cv2.CAP_PROP_FRAME_WIDTH))
height_t = int(cap_theft.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps_t = int(cap_theft.get(cv2.CAP_PROP_FPS)) or 25
total_frames_t = int(cap_theft.get(cv2.CAP_PROP_FRAME_COUNT))

print(f"Video Information:")
print(f"   - Resolution: {width_t}x{height_t}")
print(f"   - FPS: {fps_t}")
print(f"   - Total Frames: {total_frames_t}")
print(f"   - Duration: {total_frames_t/fps_t:.1f} seconds\n")

output_theft_path = "/content/output_theft_detection.mp4"
fourcc_t = cv2.VideoWriter_fourcc(*'mp4v')
out_theft = cv2.VideoWriter(output_theft_path, fourcc_t, fps_t, (width_t, height_t))

# Reset tracker for theft detection
tracker_theft = DeepSort(
    max_age=30,
    n_init=3,
    max_iou_distance=0.7,
    nn_budget=100
)

# Behavioral tracking data
behavior_data = defaultdict(lambda: {
    'positions': deque(maxlen=fps_t * 10),  # Store 10 seconds of positions
    'bboxes': deque(maxlen=fps_t * 5),      # Store 5 seconds of bboxes
    'entry_frame': None,
    'exit_frame': None,
    'risk_score': 0,
    'alerts': [],
    'zone_dwell_times': defaultdict(float),
    'last_zone': None,
    'zone_entry_time': None,
    'direction_changes': 0,
    'last_direction': None,
    'speeds': deque(maxlen=fps_t * 3),
    'suspicious_poses': 0,
})

frame_count_t = 0
alert_log = []
high_risk_snapshots = []

print("Starting theft detection analysis...")
print("Analyzing behavioral patterns...\n")

processing_start_t = time.time()

while cap_theft.isOpened():
    ret, frame = cap_theft.read()
    if not ret:
        break

    frame_count_t += 1
    current_frame_time = frame_count_t / fps_t

    if frame_count_t % 30 == 0:
        progress = (frame_count_t / total_frames_t) * 100
        elapsed = time.time() - processing_start_t
        fps_processing = frame_count_t / elapsed if elapsed > 0 else 0
        eta = (total_frames_t - frame_count_t) / fps_processing if fps_processing > 0 else 0

        print(f"Progress: {progress:.1f}% | Frame {frame_count_t}/{total_frames_t} | "
              f"Speed: {fps_processing:.1f} fps | ETA: {format_duration(eta)}")

    # ============================================
    # DETECTION
    # ============================================
    results = model.predict(
        frame,
        conf=0.4,
        classes=[0],
        verbose=False
    )

    # Pose detection for suspicious behavior
    # NOTE: 'pose_model' is not defined. This will cause an error if not handled.
    # For now, commenting out to resolve the SyntaxError, a follow-up action might be needed.
    # pose_results = pose_model.predict(
    #     frame,
    #     conf=0.4,
    #     verbose=False
    # )

    detections = []
    if len(results[0].boxes) > 0:
        for box, conf in zip(results[0].boxes.xyxy, results[0].boxes.conf):
            x1, y1, x2, y2 = box.cpu().numpy()
            confidence = float(conf)
            detections.append(([x1, y1, x2, y2], confidence, "person"))

    # ============================================
    # TRACKING
    # ============================================
    tracks = tracker_theft.update_tracks(detections, frame=frame)

    for track in tracks:
        if not track.is_confirmed():
            continue

        track_id = track.track_id
        ltrb = track.to_ltrb()
        x1, y1, x2, y2 = map(int, ltrb)

        center_x = (x1 + x2) / 2
        center_y = (y1 + y2) / 2
        bbox_width = x2 - x1
        bbox_height = y2 - y1

        person_data = behavior_data[track_id]

        # Record entry frame
        if person_data['entry_frame'] is None:
            person_data['entry_frame'] = frame_count_t

        person_data['exit_frame'] = frame_count_t

        # ============================================
        # BEHAVIORAL ANALYSIS
        # ============================================
        # Store position history
# ============================================
# AGE/GENDER ESTIMATION (every ~30 frames per track)
# ============================================
        if (frame_count_t - person_data["last_age_gender_frame"]) >= 30:
            person_roi = frame[y1:y2, x1:x2]
            if person_roi.size > 0:
                faces = detect_faces(person_roi, conf_threshold=0.7)
                if faces:
                    fx1, fy1, fx2, fy2 = max(faces, key=lambda b: (b[2]-b[0])*(b[3]-b[1]))
                    face_img = person_roi[fy1:fy2, fx1:fx2]
                    if face_img.size > 0:
                        age, gender = estimate_age_gender(face_img)
                        person_data["age"] = age
                        person_data["gender"] = gender
                        person_data["last_age_gender_frame"] = frame_count_t

        person_data['positions'].append((center_x, center_y, current_frame_time))
        person_data['bboxes'].append((bbox_width, bbox_height))

        # Reset risk score for this frame
        current_risk = 0
        frame_alerts = []

        # ============================================
        # 1. LOITERING DETECTION
        # ============================================
        if len(person_data['positions']) >= fps_t * 5:  # At least 5 seconds of data
            positions_array = np.array([(p[0], p[1]) for p in person_data['positions']])
            position_variance = np.var(positions_array, axis=0).sum()

            if position_variance < THEFT_CONFIG['zone_pixels']:
                time_span = person_data['positions'][-1][2] - person_data['positions'][0][2]
                if time_span >= THEFT_CONFIG['loitering_threshold']:
                    current_risk += RISK_WEIGHTS['loitering']
                    frame_alerts.append('LOITERING')

        # ============================================
        # 2. ERRATIC MOVEMENT DETECTION
        # ============================================
        if len(person_data['positions']) >= 3:
            recent_positions = list(person_data['positions'])[-fps_t:]  # Last second
            if len(recent_positions) >= 3:
                directions = []
                for i in range(len(recent_positions) - 1):
                    dx = recent_positions[i+1][0] - recent_positions[i][0]
                    dy = recent_positions[i+1][1] - recent_positions[i][1]
                    angle = math.atan2(dy, dx)
                    directions.append(angle)

                if len(directions) >= 2:
                    direction_changes = sum(1 for i in range(len(directions)-1)
                                          if abs(directions[i+1] - directions[i]) > math.pi/4)

                    if direction_changes >= THEFT_CONFIG['direction_changes_threshold']:
                        current_risk += RISK_WEIGHTS['erratic_movement']
                        frame_alerts.append('ERRATIC_MOVEMENT')

        # ============================================
        # 3. QUICK EXIT DETECTION
        # ============================================
        if person_data['entry_frame'] is not None:
            time_in_shop = (frame_count_t - person_data['entry_frame']) / fps_t

            # Check if near exit (bottom of frame or side edges)
            near_exit = (y2 > height_t * 0.8) or (x1 < width_t * 0.1) or (x2 > width_t * 0.9)

            if near_exit and time_in_shop <= THEFT_CONFIG['quick_exit_threshold']:
                current_risk += RISK_WEIGHTS['quick_exit']
                frame_alerts.append('QUICK_EXIT')

        # ============================================
        # 4. LOOKING AROUND (HEAD MOVEMENT)
        # ============================================
        if len(person_data['bboxes']) >= fps_t * 2:  # 2 seconds of data
            aspect_ratios = [w/h for w, h in person_data['bboxes'] if h > 0]
            if len(aspect_ratios) > 5:
                aspect_variance = np.var(aspect_ratios)
                if aspect_variance >= THEFT_CONFIG['looking_around_threshold']:
                    current_risk += RISK_WEIGHTS['looking_around']
                    frame_alerts.append('LOOKING_AROUND')

        # ============================================
        # 5. HIGH SPEED NEAR EXIT
        # ============================================
        if len(person_data['positions']) >= 2:
            last_pos = person_data['positions'][-1]
            prev_pos = person_data['positions'][-2]

            distance = math.sqrt((last_pos[0] - prev_pos[0])**2 + (last_pos[1] - prev_pos[1])**2)
            time_diff = last_pos[2] - prev_pos[2]

            if time_diff > 0:
                # Convert pixels/second to approximate m/s (rough estimate)
                speed = (distance / time_diff) / 100  # Assuming ~100 pixels = 1 meter
                person_data['speeds'].append(speed)

                near_exit = (y2 > height_t * 0.7)
                if speed > THEFT_CONFIG['high_speed_threshold'] and near_exit:
                    current_risk += RISK_WEIGHTS['high_speed']
                    frame_alerts.append('HIGH_SPEED')

        # ============================================
        # 6. SUSPICIOUS POSE DETECTION
        # ============================================
        # Check if person is bending (bbox height suddenly decreases)
        if len(person_data['bboxes']) >= fps_t:
            recent_heights = [h for w, h in list(person_data['bboxes'])[-fps_t:]]
            avg_height = np.mean(recent_heights)
            current_height = bbox_height

            # If current height is 70% or less of average, person is bending
            if current_height < avg_height * 0.7:
                current_risk += RISK_WEIGHTS['suspicious_pose']
                frame_alerts.append('SUSPICIOUS_POSE')
                person_data['suspicious_poses'] += 1

        # ============================================
        # UPDATE RISK SCORE
        # ============================================
        person_data['risk_score'] = max(person_data['risk_score'], current_risk)

        if frame_alerts:
            person_data['alerts'].extend(frame_alerts)

            # Log high-risk events
            if current_risk >= 80:
                alert_entry = {
                    'frame': frame_count_t,
                    'track_id': track_id,
                    'risk_score': current_risk,
                    'alerts': frame_alerts,
                    'timestamp': datetime.now().strftime("%H:%M:%S.%f")[:-3]
                }
                alert_log.append(alert_entry)

                # Save snapshot
                snapshot = frame[max(0, y1-20):min(height_t, y2+20),
                               max(0, x1-20):min(width_t, x2+20)].copy()
                high_risk_snapshots.append({
                    'track_id': track_id,
                    'frame': frame_count_t,
                    'snapshot': snapshot,
                    'risk_score': current_risk
                })

        # ============================================
        # VISUALIZATION
        # ============================================
        risk_color = get_risk_color(current_risk)
        risk_label_text = get_risk_label(current_risk)

        # Draw bounding box
        thickness = 3 if current_risk >= 80 else 2
        cv2.rectangle(frame, (x1, y1), (x2, y2), risk_color, thickness)

        # Prepare label
        age_text = person_data.get("age")
        gender_text = person_data.get("gender")
        if age_text and gender_text:
            label = f"ID{track_id} | Risk:{current_risk} | {risk_label_text} | {gender_text}, {age_text}"
        else:
            label = f"ID{track_id} | Risk:{current_risk} | {risk_label_text}"

        # Add alerts to label
        if frame_alerts:
            alerts_text = ", ".join(frame_alerts[:2])  # Show max 2 alerts
            label = f"{label}\n{alerts_text}"

        # Calculate label size
        label_lines = label.split('\n')
        max_label_width = 0
        total_label_height = 0

        for line in label_lines:
            (lw, lh), _ = cv2.getTextSize(line, cv2.FONT_HERSHEY_SIMPLEX, 0.5, 2)
            max_label_width = max(max_label_width, lw)
            total_label_height += lh + 5

        # Draw label background
        label_y_start = max(y1 - total_label_height - 10, 10)
        cv2.rectangle(
            frame,
            (x1, label_y_start),
            (x1 + max_label_width + 10, y1),
            risk_color,
            -1
        )

        # Draw label text
        current_y = label_y_start + 15
        for line in label_lines:
            cv2.putText(
                frame,
                line,
                (x1 + 5, current_y),
                cv2.FONT_HERSHEY_SIMPLEX,
                0.5,
                (255, 255, 255),
                2
            )
            current_y += 20

        # Draw movement trail for high-risk individuals
        if current_risk >= 60 and len(person_data['positions']) >= 2:
            points = np.array([(int(p[0]), int(p[1])) for p in person_data['positions']], dtype=np.int32)
            cv2.polylines(frame, [points], False, risk_color, 2)

    # Draw frame info
    info_text = f"Frame: {frame_count_t}/{total_frames_t} | Active Tracks: {len(tracks)}"
    cv2.putText(frame, info_text, (10, height_t - 20),
                cv2.FONT_HERSHEY_SIMPLEX, 0.6, (255, 255, 255), 2)

    out_theft.write(frame)

cap_theft.release()
out_theft.release()

processing_time_t = time.time() - processing_start_t

print("\n" + "=" * 70)
print("✅ THEFT DETECTION COMPLETE!")
print("=" * 70)
print(f"⏱️  Processing time: {format_duration(processing_time_t)}")
print(f"🎬 Total frames analyzed: {frame_count_t}")
print(f"👥 People tracked: {len(behavior_data)}")
print(f"🚨 High-risk alerts: {len(alert_log)}")
print(f"📸 Snapshots saved: {len(high_risk_snapshots)}")
print(f"📁 Output saved to: {output_theft_path}\n")
```
